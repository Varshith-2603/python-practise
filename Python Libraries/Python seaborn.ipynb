```python
#Stripplot()
import seaborn as sns
import matplotlib.pyplot as plt
x = ['sun','mon','fri','sat','tue','wed','thu']
y = [5,6.7,4,6,2,4.9,1.8]
ax = sns.stripplot(x=x,y=y)#use the x list for x axis and y list for y axis
ax.set(xlabel="Days",ylabel="Amount Spent")#setting up the names of x and y axes
plt.title("Daily spending(custom data)")
plt.show()
```

#PLOTTING WITH SEABORN AND MATPLOTLIB:-Matplotlib and Seaborn are two of the most powerful Python libraries for data visualization.
-Matplotlib provides a low-level, flexible approach to plotting
-Seaborn simplifies the process by offering built-in themes and functions for common plots.

```python
#Basic plotting using matplotlib:-
import matplotlib.pyplot as plt
plt.plot([0,1],[10,11],label='Line 1')
plt.plot([0,1],[11,10],label='Line 2')
plt.scatter([0,1],[10.5,10.5],color='blue',marker='o',label='Dots')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Simple line and dot plot")
plt.legend()
plt.show()
```

```python
#Enhancing matplotlib with seaborn styles:-
#Applying seaborn styles to matplotlib plots
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="darkgrid")
x = [1,2,3,4,5]
y = [10,12,15,18,22]
plt.plot(x,y,marker='o',linestyle='-',color='blue',label='Trend')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Matplotlib plot with seaborn theme")
plt.legend()
plt.show()
```

```python
#Customizing a seaborn plot with matplotlib
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
data = pd.DataFrame({'Year':[2018,2019,2020,2021,2022],'Sales':[100,150,200,250,300]})
plt.figure(figsize=(8,5))
sns.lineplot(x="Year",y="Sales",data=data,marker='o')
plt.title("Yearly sales growth",fontsize=14,fontweight='bold')
plt.xlabel("Year",fontsize=12)
plt.ylabel("Total Sales",fontsize=12)
plt.xticks(rotation=30)
plt.grid(True,linestyle='--')
plt.show()
```

```python
#Overlaying seaborn and matplolib plots
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
x = np.linspace(0,10,20)
y = np.sin(x)
plt.figure(figsize=(8,5))
sns.lineplot(x=x,y=y,color='blue',label='sine Wave')
plt.scatter(x,y,color='red',marker='o',label='Data Points')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Seaborn Line plot with Matplotlib scatter overlay")
plt.legend()
plt.show()
```

```python
#Enhancing seaborn with matplotlib annotations
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
data = np.random.randn(1000)
fig = plt.figure(figsize=(8,5))
sns.histplot(data,kde=True,bins=30,color='purple')
mean_value = np.mean(data)
plt.axvline(mean_value,color='red',linestyle="dashed",linewidth=2)
mean_text = "Mean: " + str(round(mean_value,2))
plt.text(mean_value+0.1,50,mean_text,color='red')
plt.title("Distribution with seaborn and matplotlib customization")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

CUSTOMIZING SEABORN PLOTS:- Control appearance and style in Seaborn.
-To modify themes,adjust colors and tailor plot.

SEABORN STYLE AND COLOR:-ways of styling the different kinds of plots in seaborn.
The ways of styling themes are(colors):- *white,*dark,*whitegrid,*darkgrid,*ticks.

```python
#Set the background to be white:-
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.set_style('white')
sns.countplot(x='sex',data=tips)
```

```python
#Set the background to ticks:-
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.set_style('ticks')
sns.countplot(x='sex',data=tips,hue='smoker',palette='deep')
print(tips.head())
```

#REMOVING AXES SPINES:-The despine() is a function that removes the spines from the right and upper portion of the plot by default.
-sns.despine(left = True) helps remove the spine from the left.

```python
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.countplot(x='sex',data=tips)
sns.despine()
```

#SCALE AND CONTEXT:-The set_context() allows us to override default parameters.
-This affects things like the size of the labels, lines, and other elements of the plot, but not the overall style. 
#poster,#paper,#notebook,#talk

```python
#POSTER
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.set_context('poster',font_scale=2)
sns.countplot(x='sex',data=tips,palette='coolwarm')
```

```python
#PAPER
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.set_context('paper',font_scale=2)
sns.countplot(x='sex',data=tips,palette='coolwarm')
```

```python
#NOTEBOOK
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.set_context("notebook",font_scale=2)
sns.countplot(x="sex",data=tips,palette='coolwarm')
```

```python
#TALK
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.set_theme()
sns.set_context("talk",font_scale=2)
sns.countplot(x='sex',data=tips,palette='coolwarm')
```

#GRIDS AND MULTI-PLOT LAYOUTS:-Seaborn’s grid-based layouts that help compare multiple visualizations at once.
-They are useful for exploring data across categories, subgroups or variable combinations.

#FacetGrid():-used to create multiple plots from the same dataset, split according to one or more categorical variables.

```python
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
sns.set_theme()
g = sns.FacetGrid(tips, col="sex")
g.map(sns.scatterplot, "total_bill", "tip")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
graph = sns.FacetGrid(data=tips, col ="sex",  hue ="day")
sns.set_theme()
graph.map(plt.scatter, "total_bill", "tip", edgecolor ="w")
plt.legend(["Thur","Fri","Sat","Sun"])
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
g = sns.FacetGrid(tips,row="smoker",col='time')
g.map(plt.hist,'total_bill',bins=15,color='green')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

g = sns.FacetGrid(tips, row="smoker", col="time")

g.map(plt.hist, "total_bill", bins=15, color="green")

g.set(xlim=(0, 55), xticks=[0, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55])

plt.tight_layout()
plt.show()
```
