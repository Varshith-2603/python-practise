PLOTTING HISTOGRAMIN PYTHON USING MATPLOTLIB:-A histogram is used to visualize the distribution of numerical data by grouping values into intervals called bins.
-It helps identify the frequency of data values, making it easier to observe patterns, spread and the overall distribution of a dataset.


```python
#Sample histogram:-
import matplotlib.pyplot as plt
data = [2,3,3,4,5,5,5,6,7,8]
plt.hist(data)
plt.show()
```

Creating a basic histogram:-A basic histogram groups numerical data into bins and displays the frequency of values in each interval.

```python
import matplotlib.pyplot as plt
import numpy as np
data = np.random.randn(1000)
plt.hist(data,bins=30,color='skyblue',edgecolor='grey')
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Basic Histogram")
plt.show()
```

```python
#Histogram with kde:-
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
data = np.random.randn(1000)
sns.histplot(data,bins=30,kde=True,color='lightgreen',edgecolor='red')
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Histogram with kde curve")
plt.show()
```

```python
#Multiple histograms using subplots:-
import matplotlib.pyplot as plt
import numpy as np
data1 = np.random.randn(1000)
data2 = np.random.normal(loc=3,scale=1,size=1000)
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(12,4))
axes[0].hist(data1,bins=30,color='Yellow',edgecolor='black')
axes[0].set_title("Histogram 1")
axes[1].hist(data2,bins=30,color="pink",edgecolor='black')
axes[1].set_title("Histogram 2")
for ax in axes:
    ax.set_xlabel("Values")
    ax.set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
#Stacked histogram:-
import matplotlib.pyplot as plt
import numpy as np
data1 = np.random.randn(1000)
data2 = np.random.normal(loc=3,scale=1,size=1000)#mean/centre is approx:3,standard deviation:1,size=1000 values
plt.hist([data1,data2],bins=30,stacked=True,color=['cyan','purple'],edgecolor='black')
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Stacked histogram")
plt.legend(['Dataset 1','Dataset 2'])
plt.show()
```

PLOT A PIECHART IN PYTHON USING MATPLOTLIB:-A pie chart is a circular graph used to represent data as slices of a circle, where each slice corresponds to a category's proportion of the whole.
-It is commonly used to compare the percentage or share of different categories in a dataset.

```python
import matplotlib.pyplot as plt
labels = ["A","B","C","D"]
values = [30,25,20,25]
plt.pie(values,labels=labels,autopct="%1.0f%%")
plt.title("Simple Pie Chart")
plt.show()
```

```python
#Customizing a pie chart
import matplotlib.pyplot as plt
import numpy as np
cars = ["AUDI","BMW","FORD","TESLA","JAGUAR","MERCEDES"]
data = [23,17,35,29,12,41]
explode = [0.1,0.0,0.2,0.3,0.0,0.0]
colors = ("orange","cyan","brown","grey","indigo","beige")
wp = {'linewidth':1,'edgecolor':"green"}
def func(pct,allvalues):
    absolute = int(pct / 100.*np.sum(allvalues))
    return "{:.1f}%\n({:d} g)".format(pct, absolute)
fig,ax = plt.subplots(figsize=(10,7))
wedges,text,autotexts = ax.pie(data,autopct=lambda pct: func(pct,data),explode=explode,labels=cars,shadow=True,colors=colors,startangle=90,wedgeprops=wp,textprops=dict(color='magenta'))
ax.legend(wedges,cars,title="Cars",loc="center left",bbox_to_anchor=(1,0,0.5,1))
plt.setp(autotexts,size=8,weight='bold')
ax.set_title("Customizing pie chart")
plt.show()
```

ADDING VALUE LABELS ON A MATPLOTLIB BAR CHART:-Methods to add value labels on a bar chart
Matplotlib provides multiple methods for adding value labels to a bar chart. We will primarily use the following two:
-Creating a Bar Chart using plt.bar()
-Adding Text Labels using plt.text()

```python
#Adding value labels on top of bars
import matplotlib.pyplot as plt
def add_labels(x,y):
    for i in range(len(x)):
        plt.text(i,y[i],y[i])

x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]
plt.bar(x,y)
add_labels(x,y)
plt.title("College admission")
plt.xlabel("Courses")
plt.ylabel("Number of admissions")
plt.show()
```

```python
#Centering value labels on top of bars:-
import matplotlib.pyplot as plt
def add_labels(x,y):
    for i in range(len(x)):
        plt.text(i,y[i],y[i],ha='center')
x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]
plt.figure(figsize=(10,5))
plt.bar(x,y)
add_labels(x,y)
plt.title("College Admission")
plt.xlabel("Courses")
plt.ylabel("Number of admissions")
plt.show()
```

```python
#Displaying value labels inside the rectangular box:-
import matplotlib.pyplot as plt
def add_labels(x,y):
    for i in range(len(x)):
        plt.text(i,y[i],y[i],ha='center',bbox=dict(facecolor='red',alpha=0.8))
x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]
plt.figure(figsize=(10,5))
plt.bar(x,y)
add_labels(x,y)
plt.title("College admission")
plt.xlabel("Courses")
plt.ylabel("Number of admissions")
plt.show()
```

```python
#Placing labels at the center of each bar:
import matplotlib.pyplot as plt
def add_labels(x,y):
    for i in range(len(x)):
        plt.text(i,y[i]//2,y[i],ha='center')
x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]
plt.figure(figsize=(10,5))
plt.bar(x,y,color='green')
add_labels(x,y)
plt.xlabel("Courses")
plt.ylabel("Number of students")
plt.show()
```

#GRIDS IN MATPLOTLIB:-Grids in Matplotlib are intersecting lines that provide a structured reference for data points, improving readability and interpretation of plots.

```python
#Addinmg grid line to plot()
import matplotlib.pyplot as plt
import numpy as np
x1 = np.linspace(0,5,100)
y1 = np.cos(2*np.pi*x1)*np.exp(-x1)
fig,(ax1,ax2) = plt.subplots(1,2,figsize=(12,5))
ax1.plot(x1,y1)
ax1.set_title("Plot Without grid")
ax2.plot(x1,y1)
ax2.set_title("Plot With Grid")
ax2.grid(True)
plt.show()
```

```python
#Customizing grid lines:-
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,2*np.pi,400)
y = np.sin(x**2)
plt.plot(x,y,color='green')
plt.grid(True,color='grey',linewidth=1.4,linestyle='-.')
plt.title("PLot with customizing grid lines")
plt.show()
```

```python
#Display lines on x-axis only:-
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,2*np.pi,400)
y = np.sin(x**2)
plt.plot(x,y,color='green')
plt.title("Display grid lines only on x-axis")
plt.grid(True,color='grey',linewidth=1.4,axis='x')
plt.show()
```

```python
#Display lines only on the y-axis:
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,2*np.pi,400)
y = np.sin(x**2)
plt.plot(x,y,color='red')
plt.title("Displaying grid lines only on y-axis")
plt.grid(True,color='grey',linestyle='-.',axis='y',linewidth=1.4)
plt.show()
```

#SUBPLOTS IN MATPLOTLIB:-Subplots in Matplotlib refer to multiple plots arranged within a single figure.

```python
#Creating a grid of subplots()
import matplotlib.pyplot as plt
import numpy as np
fig,ax = plt.subplots(3,3)
for row in ax:
    for col in row:
        col.plot(np.random.randint(0,5,5),np.random.randint(0,5,5))
plt.show()
```

```python
#Creating subplots for two different types of plots:-
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,4,6,8,10]
fig,axes = plt.subplots(1,2,figsize=(10,4))
axes[0].plot(x,y,marker='o',color='red')
axes[0].set_title("Line Plot")
axes[0].set_xlabel("X-Axis")
axes[0].set_ylabel("Y-Axis")
axes[1].scatter(x,y,marker='*',color="magenta")
axes[1].set_title("Scatter Plot")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")
plt.tight_layout()
plt.show()
```

```python
#Piecharts in subplots:-
import matplotlib.pyplot as plt
labels = ['Category 1', 'Category 2', 'Category 3']
sizes1 = np.random.randint(1, 10, 3)
sizes2 = np.random.randint(1, 10, 3)
sizes3 = np.random.randint(1, 10, 3)
fig,axes = plt.subplots(nrows=1,ncols=3,figsize=(12,4))
axes[0].pie(sizes1,labels=labels,autopct="%1.0f%%",colors=["lightcoral",'lightblue','lightgreen'])
axes[1].pie(sizes2,labels=labels,autopct="%1.0f%%",colors=['gold','lightseagreen','lightpink'])
axes[2].pie(sizes3,labels=labels,autopct="%1.0f%%",colors=['lightskyblue','lightgreen','lightcoral'])
axes[0].set_title('Pie Chart 1')
axes[1].set_title("Pie Chart 2")
axes[2].set_title("Pie Chart 3")
plt.tight_layout()
plt.show()
```

```python
#Customizing subplots using gridspec()
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np
gs = gridspec.GridSpec(2,3,width_ratios=[1,2,1],height_ratios=[2,1])
ax1 = plt.subplot(gs[0,0])
ax2 = plt.subplot(gs[0,1])
ax3 = plt.subplot(gs[0,2])
ax4 = plt.subplot(gs[1,:])
x = np.linspace(0,10,100)
y = np.sin(x)
ax1.plot(x,y,color='blue')
ax1.set_title("Line Plot:Sine function")
x = np.random.rand(30)
y2 = 3*x+np.random.randn(30)
ax2.scatter(x,y2,color='green')
ax2.set_title("Scatter Plot")
categories = ['A', 'B', 'C', 'D']
values = np.random.randint(1, 10, 4)
ax3.bar(categories, values, color='orange')
ax3.set_title('Bar Plot')
labels = ['Category 1', 'Category 2', 'Category 3']
sizes = np.random.randint(1, 10, 3)
ax4.pie(sizes,labels=labels,autopct="%1.0f%%",colors=['lightcoral', 'lightblue', 'lightgreen'])
ax4.set_title("Pie Chart")
plt.tight_layout()
plt.show()
```

#RESIZE PLOTS:-Matplotlib provides a default figure size of 6.4 inches in width and 4.8 inches in height.
-To change the size of a figure in matplotlib, we can use the figsize parameter of the plt.figure() function.

```python
#Changing figure size using figsize parameter:-
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-5,5,100)
y = x**2
plt.figure(figsize=(5,2))
plt.plot(x,y)
plt.title("Plot of y=x^2")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
#Setting Width of the Figure with set_figwidth()
import matplotlib.pyplot as plt
x = [1,3,5,7,9]
y = [5,7,2,8,4]
plt.plot(x,y)
plt.title("X vs Y")
plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.show()
```

```python
#Setting Height of Figure with set_figheight()
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [12,5,18,7,9]
plt.figure().set_figheight(5)
plt.plot(x,y)
plt.title("X vs Y",color='red')
plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.show()
```

#ADJUST PLOT TRANSPARENCY:-Changing transparency in Matplotlib plots enhances visual clarity, especially when data overlaps.
-Transparency is controlled using a value between 0 (fully transparent) and 1 (fully opaque). 

```python
#Using alpha parameter:-alpha parameter is the most direct way to control transparency when creating plots in Matplotlib.
#LINE PLOT WITH LOW ALPHA:-
import matplotlib.pyplot as plt
x = [0,5,10,15]
y = [0,1,2,3]
plt.plot(x,y,color='blue',alpha=0.3)
plt.title("Line Plot with alpha = 0.3",color='black')
plt.show()
```

```python
#Histogram with transparent bars:-
import matplotlib.pyplot as plt
import numpy as np
d = np.random.randn(1000)
plt.hist(d, bins=30, color='orange', alpha=0.4)
plt.title("Histogram with transparent bars",color='red')
plt.show()
```

#USING SET ALPHA():-set_alpha() method is useful when you want to adjust the transparency of plot elements dynamically after they have already been created.

```python
#Set alpha after line plotting:-
import matplotlib.pyplot as plt
line, = plt.plot([1,2,3],[4,5,6])
line.set_alpha(0.5)
plt.title("Alpha set using set alpha",color='magenta')
plt.show()
```

#CHANGE FONTS:-Changing fonts in Matplotlib helps customize the appearance of plots, making them more readable and visually appealing.
Fonts can be changed for titles, axis labels, legends and other text elements either individually or globally.


```python
#Using Font name:-This method lets you apply different fonts to specific parts of your plot, like axis labels or the title.
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y,color='red')
plt.xlabel("X-values",fontname='Brush Script MT',fontsize=16,color='red')
plt.ylabel("Y-values",fontname='Brush Script MT',fontsize=16,color='red')
plt.show()
```

```python
#Change font of title only:-
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,2,3,4,5])
y = x**2
plt.plot(x,y,color='green',linestyle='--')
plt.title("Plot of y=x**2",color='red',fontname="Franklin Gothic Medium",fontsize=18)
plt.xlabel("X-Values")
plt.ylabel("Y-values")
plt.show()
```

```python
#Title and Axis labels together:-
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 35]
plt.plot(x, y)
plt.title("Population Over Time", fontname="Franklin Gothic Medium", fontsize=18,color='black')
plt.xlabel("Years", fontname="Gabriola", fontsize=16,color='red')
plt.ylabel("Population (Million)", fontname="Gabriola", fontsize=16,color='red')
plt.show()
```

#SETTING TICKS LABEL FONT SIZE:-One important aspect of making plots readable and aesthetically pleasing is formatting tick labels, including adjusting their font size.
-Tick Labels: Ticks are markers indicating specific data points on an axis and labels attached to ticks help in understanding the values represented.
-Font Size: Controls the size of text in tick labels and enhances clarity and readability of plots.
